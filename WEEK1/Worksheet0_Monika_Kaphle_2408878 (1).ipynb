```python
# 1 List Manipulation 4.2
def extract_every_other(lst):
    """Return every other element."""
    return lst[::2]

print(extract_every_other([1, 2, 3, 4, 5, 6]))
```

```python
# 2 Slice sublist
def get_sublist(lst, start, end):
    """Return sublist from start to end inclusive."""
    return lst[start:end+1]

print(get_sublist([1, 2, 3, 4, 5, 6], 2, 4))
```

```python
# 3 Reverse list
def reverse_list(lst):
    """Reverse list using slicing."""
    return lst[::-1]

print(reverse_list([1, 2, 3, 4, 5]))
```

```python
# 4 Remove first and last
def remove_first_last(lst):
    """Remove first and last element."""
    return lst[1:-1]

print(remove_first_last([1, 2, 3, 4, 5]))
```

```python
# 5 first n elements
def get_first_n(lst, n):
    """Return first n elements."""
    return lst[:n]

print(get_first_n([1, 2, 3, 4, 5], 3))
```

```python
# 6 last n elements
def get_last_n(lst, n):
    """Return last n elements."""
    return lst[-n:]

print(get_last_n([1, 2, 3, 4, 5], 2))
```

```python
# 7 reverse skip
def reverse_skip(lst):
    """Return every second element from second last backwards."""
    return lst[-2::-2]

print(reverse_skip([1, 2, 3, 4, 5, 6]))
```

```python
# Nested list exercise 4.3
# 1 flatten nested list
def flatten(lst):
    """Flatten nested list."""
    result = []
    for sub in lst:
        result.extend(sub)
    return result

print(flatten([[1, 2], [3, 4], [5]]))
```

```python
# 2 Access nested element
def access_nested_element(lst, indices):
    """Access element using index list."""
    element = lst
    for i in indices:
        element = element[i]
    return element

lst = [[1,2,3],[4,5,6],[7,8,9]]
print(access_nested_element(lst, [1,2]))
```

```python
# 3 Sum nested
def sum_nested(lst):
    """Sum elements in nested list."""
    total = 0
    for item in lst:
        if isinstance(item, list):
            total += sum_nested(item)
        else:
            total += item
    return total

print(sum_nested([[1,2],[3,[4,5]],6]))
```

```python
# 4 remove element
def remove_element(lst, elem):
    """Remove element from nested list."""
    result = []
    for item in lst:
        if isinstance(item, list):
            result.append(remove_element(item, elem))
        elif item != elem:
            result.append(item)
    return result

print(remove_element([[1,2],[3,2],[4,5]], 2))
```

```python
# 5 find max
def find_max(lst):
    """Find maximum in nested list."""
    max_val = float("-inf")
    for item in lst:
        if isinstance(item, list):
            max_val = max(max_val, find_max(item))
        else:
            max_val = max(max_val, item)
    return max_val

print(find_max([[1,2],[3,[4,5]],6]))
```

```python
# 6 count occurance
def count_occurrences(lst, elem):
    """Count element in nested list."""
    count = 0
    for item in lst:
        if isinstance(item, list):
            count += count_occurrences(item, elem)
        elif item == elem:
            count += 1
    return count

print(count_occurrences([[1,2],[2,3],[2,4]], 2))
```

```python
# 7 deep flatten
def deep_flatten(lst):
    """Flatten list regardless of depth."""
    result = []
    for item in lst:
        if isinstance(item, list):
            result.extend(deep_flatten(item))
        else:
            result.append(item)
    return result

print(deep_flatten([[[1,2],[3,4]],[[5,6],[7,8]]]))
```

```python
# 8 nested average
def average_nested(lst):
    """Return average of nested list."""
    flat = deep_flatten(lst)
    return sum(flat) / len(flat)

print(average_nested([[1,2],[3,4],[5,6]]))
```

```python
import numpy as np

# Define matrices A and B
A = np.array([[3, 4],
              [7, 8]])
B = np.array([[5, 3],
              [2, 1]])
```

```python
#1
# Compute inverse of A
A_inv = np.linalg.inv(A)

# Multiply A with its inverse
identity = np.dot(A, A_inv)

print("A · A⁻¹:\n", identity)
```

```python
#2
AB = np.dot(A, B)
BA = np.dot(B, A)

print("A·B:\n", AB)
print("B·A:\n", BA)
print("Are A·B and B·A equal?", np.array_equal(AB, BA))
```

```python
#3
AB_T = np.dot(A, B).T
B_T_A_T = np.dot(B.T, A.T)

print("(A·B)ᵀ:\n", AB_T)
print("Bᵀ·Aᵀ:\n", B_T_A_T)
print("Are they equal?", np.array_equal(AB_T, B_T_A_T))
```

```python
#4
# Coefficient matrix
coeff = np.array([[2, -3, 1],
                  [1, -1, 2],
                  [3, 1, -1]])

# Constant matrix
const = np.array([-1, -3, 9])
```

```python
#using np.invert # Compute inverse of coefficient matrix
coeff_inv = np.linalg.inv(coeff)

# Solve for X
X = np.dot(coeff_inv, const)

print("Solution [x, y, z]:", X)
```
